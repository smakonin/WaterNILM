```python
# finds all the timestamps where load switches on/off igonring off periods of length < threshold
def switchTimes (load, threshold):
    times = [0]
    zero = True # Assumes we start and end with an off period
    for i,x in enumerate(load):
        if zero and x != 0 :
            zero = False
            times.append(i)
        elif not zero and x == 0:
            zero = True
            times.append(i)
    # Remove all off periods less than threshold min
    # Incorrectly assumes first and last periods are greater than threshold min
    filtered = sum([(off, on) for off, on in zip(times[::2], times[1::2]) if on - off > threshold], ())
    return filtered
```

```python
# represent a discretized load as a series of spans (measurement, duration, start, end)
def spans(load):
    res = []
    cur = load[0]
    start = 0
    for index, val in enumerate(load):
        if val != cur:
            res.append((cur, index - start, start, index))
            cur = val
            start = index
    res.append((cur, len(load) - start, start, len(load)))
    return res
```

```python
# extract events from discretized load
# mix of "switchTimes" and "spans"
# returns list of ([bounds], [measurements])
def events(load, threshold):
    res = []
    bounds = [0]
    measurements = []
    cur = load[0]
    for index, val in enumerate(load):
        if val != cur:
            if cur == 0 and index - bounds[-1] > threshold:
                res.append((bounds, measurements))
                bounds = [index]
                measurements = []
            else:
                bounds.append(index)
                measurements.append(cur)
            cur = val
    return res
```
