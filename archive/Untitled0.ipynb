```python
# finds all the timestamps where load switches on/off igonring off periods of length < threshold
def switchTimes (load, threshold):
    times = [0]
    zero = True # Assumes we start and end with an off period
    for i,x in enumerate(load):
        if zero and x != 0 :
            zero = False
            times.append(i)
        elif not zero and x == 0:
            zero = True
            times.append(i)
    # Remove all off periods less than threshold min
    # Incorrectly assumes first and last periods are greater than threshold min
    filtered = sum([(off, on) for off, on in zip(times[::2], times[1::2]) if on - off > threshold], ())
    return filtered
```

```python
timesDWE = switchTimes (DWE, 2)
timesCWE = switchTimes (CWE, 5)
timesWHW = switchTimes (WHW, 0)
```

```python
# spans of (start, end) off/on times
offDWE = zip(timesDWE[::2], timesDWE[1::2])
onDWE = zip(timesDWE[1::2], timesDWE[2::2])
offCWE = zip(timesCWE[::2], timesCWE[1::2])
onCWE = zip(timesCWE[1::2], timesCWE[2::2])
offWHW = zip(timesWHW[::2], timesWHW[1::2])
onWHW = zip(timesWHW[1::2], timesWHW[2::2])
# adding length --> (start, end, length)
offDWE = [(x, y, y - x) for x, y in offDWE]
onDWE = [(x, y, y - x) for x, y in onDWE]
offCWE = [(x, y, y - x) for x, y in offCWE]
onCWE = [(x, y, y - x) for x, y in onCWE]
offWHW = [(x, y, y - x) for x, y in offWHW]
onWHW = [(x, y, y - x) for x, y in onWHW]

print(len(onDWE))
# remove ones less than 100 min
onDWE = [(x,y,z) for (x,y,z) in onDWE if z >= 100 and z < 140]

print(len(onDWE))
```
