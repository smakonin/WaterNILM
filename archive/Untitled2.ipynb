```python
# represent a discretized load as a series of spans (measurement, duration)
def spans(load):
    res = []
    cur = load[0]
    length = 0
    for val in load:
        if val != cur:
            res.append((cur, length))
            cur = val
            length = 0
        length += 1
    return res
```

```python
# get the spans both in terms of actual values and bins (same thing really)
DWE_ds = spans(DWE_d)
DWE_bs = spans(DWE_b)
CWE_ds = spans(CWE_d)
CWE_bs = spans(CWE_b)
```
