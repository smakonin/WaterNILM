```python
import numpy as np

%matplotlib inline
import matplotlib.pyplot as plt
import matplotlib.ticker as tck
```

```python
DWE = np.genfromtxt("../AMPds/electricity/DWE.csv", delimiter = ",", names = True, skip_footer = 900750)
WHW = np.genfromtxt("../AMPds/water/WHW.csv", delimiter = ",", names = True, skip_footer = 900750)
```

```python
# finds nonzero slices in data, ignoring sequences of zeros of length <=skip
# returns list of (start, end) for convenience
def samples(data, skip=0):
    indicies = np.nonzero(data)[0]
    return [(x[0], x[-1]+1) for x in np.split(indicies, np.where(np.diff(indicies)>skip+1)[0]+1)]
```

```python
DWsamples = samples(DWE["S"])
```

```python
for e,(start,end) in enumerate(DWsamples):
    print("# Sample %i" % e)
    for i in range(start,end):
        if WHW["AVG_RATE"][i] != 0:
            print("DWW[%i] =" % i, WHW["AVG_RATE"][i])
```

```python
for e,(start,end) in enumerate(DWsamples):
    fig,ax = plt.subplots(figsize=(20,4))
    fig.suptitle("Sample %i" % e)
    ax.plot(range(start,end),WHW["AVG_RATE"][start:end],marker='o')
    ax.plot(range(start,end),DWE["I"][start:end],marker='o')
    ax.set_ylim([0,8])
    lo = np.around(start - 5, -1)
    hi = np.around(end + 5, -1)
    ax.set_xlim([lo,hi])
    ax.set_xticks(np.arange(lo, hi, 10))
    ax.xaxis.set_major_formatter(tck.ScalarFormatter(useOffset=False))
    ax.minorticks_on()
    ax.set_xticks(np.arange(lo, hi, 2), minor = True)
    ax.grid()
    ax.grid(which='minor', axis='x')
    plt.show()
```
